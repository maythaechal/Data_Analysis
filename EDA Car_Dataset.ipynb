### EDA (Exploratpry Data Analysis)

* exploring and understanding your data before performing formal analysis
* Without EDA: may get wrong conclusions
* Understand  the Data
* Check Missing Values
* Data Visualization
  * Histogram -> to see distribution
  * Scatter plot -> to see relationships
  * Box plot -> to detect outliers

```python
car_df=pd.read_csv("D:\YIU\YIU\LevelFive\BPS\Cleaned_Data\cleaned_Car.csv")
```

```python
car_df.head()
```

```python
numerical_features = ['price', 'stroke', 'enginesize', 'curbweight', 'compressionratio', 'highwaympg','peakrpm','citympg']
correlation_matrix = car_df[numerical_features].corr()
```

```python
correlation_matrix
```

```python
plt.figure(figsize=(6,4))
sns.heatmap(correlation_matrix, annot=True, cmap="Greens", linewidths=0.5)
plt.title('Heatmap of Numerical Features with Correlation Values')
plt.xticks(rotation=45)
plt.yticks(rotation=45)
plt.show()
```

```python
sns.barplot(x='enginetype', y='price', data=car_df)
plt.show()
```

```python
sns.barplot(x='carbody', y='price', data=car_df)
plt.show()
```

```python
carbody_counts = car_df['carbody'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(carbody_counts, labels=carbody_counts.index, autopct='1.1f%%', startangle=45)
plt.title('Distribution of Car Bodies')
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.histplot(car_df['price'], bins=20, kde=True, color='Green')
plt.title("Distribution of Car Prices with KDE")
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(10,6))
sns.boxplot(x='carbody', y='price', data=car_df)
plt.title('Car Prices by Car Body Type')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.figure(figsize=(8,6))
plt.scatter(car_df['citympg'], car_df['price'], color='purple')
plt.title('Horsepower vs Price')
plt.xlabel('Horsepower')
plt.ylabel('Price')
plt.show()
```

```python
sns.pairplot(car_df[['price',  'enginesize', 'citympg', 'highwaympg']])
plt.show()
```
